```python
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate

def build_chat_prompt(role_type, topic_name):
    system_template = "You are a helpful {role}."
    system_message = SystemMessagePromptTemplate.from_template(system_template)

    user_template = "Please tell me about {topic}."
    user_message = HumanMessagePromptTemplate.from_template(user_template)

    chat_prompt = ChatPromptTemplate.from_messages([system_message, user_message])
    return chat_prompt.format_prompt(role=role_type, topic=topic_name).to_messages()
messages = build_chat_prompt(role_type='teacher', topic_name='Neural Networks')
```

```python
def validate_inputs(audience, tone):
    valid_audiences = ['beginner', 'intermediate', 'expert']
    valid_tones = ['formal', 'casual', 'fun']

    final_audience = audience if audience in valid_audiences else 'beginner'
    final_tone = tone if tone in valid_tones else 'formal'

    return final_audience, final_tone

def generate_prompt(topic, audience, tone, style):
    v_audience, v_tone = validate_inputs(audience, tone)

    styles = {
        "teaching": "Explain {topic} for {audience} in a {tone} teaching style",
        "interview": "Ask interview questions about {topic} for an {audience} level in a {tone} tone",
        "storytelling": "Explain {topic} for {audience} in a {tone} storytelling style" [cite: 104]
    }

    selected_template = styles.get(style, styles["teaching"])
    prompt_temp = PromptTemplate(input_variables=["topic", "audience", "tone"], template=selected_template)

    return prompt_temp.format(topic=topic, audience=v_audience, tone=v_tone)
```

```python
from langchain_core.prompts import PromptTemplate

reusable_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Summarize the core concepts of {topic} for an {audience} audience in a {tone} manner."
)
test_inputs = [
    {"topic": "Quantum Computing", "audience": "expert", "tone": "academic"},
    {"topic": "Cooking Pasta", "audience": "beginner", "tone": "friendly"},
    {"topic": "Stock Market Volatility", "audience": "intermediate", "tone": "professional"},
    {"topic": "Photosynthesis", "audience": "kids", "tone": "fun"},
    {"topic": "Cybersecurity Best Practices", "audience": "employees", "tone": "urgent"}
]
print("--- Template Reusability Test Results ---")

for data in test_inputs:
    output = reusable_template.format(
        topic=data["topic"],
        audience=data["audience"],
        tone=data["tone"]
    )
    print(output)
```
