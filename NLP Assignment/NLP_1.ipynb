## **Task 1**

1. What is the difference between "Love" and "love" in NLP?

These are completely different because of the capital (L). If we don’t lowercase our text, the model treats them as two unrelated wordswhich bloats the vocabulary and makes it harder for the system to realize they actually mean the same thing.

2. What happens if stopwords are not removed?

If you leave in words like (the) or (is) they act as "noise" that can drown out the actual meaning of the text. Because these words appear so frequently they take up unnecessary processing power and can trick a model into thinking they are more important than the actual keywords.

3. Provide two real-world scenarios where removing stopwords can be harmful.

In sentiment analysis removing (not) can turn (not happy) into (happy) which completely ruins the prediction. It also breaks search engines if you strip the stopwords from a famous quote like To be or not to be you're left with almost nothing to search for.

4. Explain the difference between stemming and lemmatization.

Stemming is a quick-and-dirty method that just chops off the ends of words often leaving you with non-words like (comput) Lemmatization is much smarter because it uses a dictionary to find the actual root word like turning (was) into (be) ensuring the result actually makes sense.

# **Task 2**

```python
import re
import string

def preprocess_text(text):
    if not text or not isinstance(text, str):
        return []

    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'(.)\1{2,}', r'\1', text)
    text = text.translate(str.maketrans('', '', string.punctuation))
    text = " ".join(text.split())
    tokens = text.split()
    meaningful_exceptions = {"no", "not"}
    cleaned_tokens = [word for word in tokens if len(word) > 2 or word in meaningful_exceptions]

    return cleaned_tokens

test_sentence = "Visit  http://example.com now! Nooooo this is baaad!!! I have 2 dogs."
result = preprocess_text(test_sentence)

print(f"Original: {test_sentence}")
print(f"Cleaned Tokens: {result}")
```

# **Task 3**

```python
sample_inputs = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]
for text in sample_inputs:
    tokens = preprocess_text(text)
    cleaned_sentence = " ".join(tokens)
    print(f"{str(tokens):<40} | {cleaned_sentence}")
```

# **Task 4**

```python
print(f"{'Processed Sentence':<35} | {'Total':<6} | {'Unique':<6} | {'Avg Length':<10}")
print("-" * 75)

sentence_stats = []

for text in sample_inputs:
    tokens = preprocess_text(text)

    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))
    avg_length = sum(len(t) for t in tokens) / total_tokens if total_tokens > 0 else 0

    cleaned_sent = " ".join(tokens)
    sentence_stats.append({
        "original": text,
        "clean": cleaned_sent,
        "total": total_tokens,
        "noise_score": len(text) - len(cleaned_sent)
    })

    print(f"{cleaned_sent[:34]:<35} | {total_tokens:<6} | {unique_tokens:<6} | {avg_length:<10.2f}")
```

# **Task 5**

```python
from collections import Counter

all_tokens = []
for text in sample_inputs:
    tokens = preprocess_text(text)
    all_tokens.extend(tokens)
word_freq = Counter(all_tokens)
top_10 = word_freq.most_common(10)
least_5 = sorted(word_freq.items(), key=lambda x: x[1])[:5]
print("--- Frequency Analysis Results ---")

print("\nTop 10 Most Frequent Words:")
print(f"{'Word':<15} | {'Frequency'}")
print("-" * 30)
for word, freq in top_10:
    print(f"{word:<15} | {freq}")

print("\nTop 5 Least Frequent Words:")
print(f"{'Word':<15} | {'Frequency'}")
print("-" * 30)
for word, freq in least_5:
    print(f"{word:<15} | {freq}")
```

# **Task 6**

```python
import json
def full_pipeline(text_list):

    if not isinstance(text_list, list):
        return {"tokens": [], "clean_sentences": []}

    all_tokens = []
    all_clean_sentences = []

    for text in text_list:

        tokens = preprocess_text(text)
        all_tokens.append(tokens)
        all_clean_sentences.append(" ".join(tokens))

    return {
        "tokens": all_tokens,
        "clean_sentences": all_clean_sentences
    }
pipeline_output = full_pipeline(sample_inputs)
print(json.dumps(pipeline_output, indent=4))
```
